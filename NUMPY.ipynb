```python
#slicing - selecting a range of elements
# arra[start,end] start to end -1
import numpy as np
arr1=np.array([1,2,3,4,5,6,7])
print(arr1[1:3])
print(arr1[:5])
print(arr1[::2])#outputs every second element
print(arr1[::-1])#prints array in reverse order
print(arr1[[0,1,2,6]])#fancy indexing
print(arr1[arr1>4]) # boolean masking



```

```python
import pandas as pd
import numpy as np
df=pd.read_csv('/content/indian_employee_data_300 (1).csv')




df['Salary (INR)'].fillna(df['Salary (INR)'].mean(),inplace=True)
df['Performance Rating'].fillna(df['Performance Rating'].median(),inplace=True)
df.replace([np.inf,-np.inf],np.nan, inplace=True)


# remove duplicate record
df.drop_duplicates(inplace=True)
df['Salary (INR)']=np.where(df['Salary (INR)']<0 , df['Salary (INR)'].mean(),df['Salary (INR)'])
salary_mean=df['Salary (INR)'].mean()
salary_std=df['Salary (INR)'].std()
lower_bound=salary_mean-(3*salary_std)
upper_bound=salary_mean+(3*salary_std)

df=df[(df['Salary (INR)']>=lower_bound) & (df['Salary (INR)']<=upper_bound)]
df.to_csv('clean_indianemployee_data.csv',index=False)
print('Data cleaned')

print()
print("Missing value in file")
print(df.isnull().sum())

print()
print(df.head())
```
